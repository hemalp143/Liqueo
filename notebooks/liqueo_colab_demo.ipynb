```
from liqueo.core import Document, KnowledgeBase, generate_doc_id
from liqueo.embeddings import EmbeddingsManager
from liqueo.recommender import RecommendationEngine
from liqueo.synthesizer import KnowledgeSynthesizer
from datetime import datetime
print('✓ Imports complete')
```

## Load Sample Data

```
kb = KnowledgeBase()

docs = [
    {'title': 'SaaS Acquisition', 'industry': 'Technology', 'type': 'M&A', 'value': 2.5, 'duration': 8},
    {'title': 'Retail Restructuring', 'industry': 'Retail', 'type': 'Restructuring', 'value': 1.8, 'duration': 12},
    {'title': 'Bank Reorganization', 'industry': 'Financial Services', 'type': 'Restructuring', 'value': 3.2, 'duration': 10}
]

for d in docs:
    doc = Document(
        id=generate_doc_id(d['title'], datetime.now()),
        title=d['title'],
        industry=d['industry'],
        transaction_type=d['type'],
        engagement_value=d['value'],
        duration_months=d['duration'],
        content=f"Consulting engagement: {d['title']}"
    )
    kb.add_document(doc)

print(f'✓ Loaded {len(kb.list_documents())} documents')
```

## Search

```
embeddings = EmbeddingsManager(kb)
results = embeddings.semantic_search('technology infrastructure optimization', top_k=3)
print(f'Found {len(results)} results:\n')
for i, r in enumerate(results, 1):
    print(f'{i}. {r.document.title} ({r.similarity_score:.0%})')
```

## Recommendations

```
recommender = RecommendationEngine(embeddings)
recs = recommender.recommend_similar_engagements('fintech cost optimization', top_k=3)
print(f'Recommendations:\n')
for i, rec in enumerate(recs, 1):
    print(f'{i}. {rec.reference_document.title} ({rec.relevance_score:.0%})')
```

## Synthesis

```
synthesizer = KnowledgeSynthesizer(kb, recommender)
synthesis = synthesizer.synthesize_recommendations('healthcare digital transformation', top_k=2)
print(synthesis)
```
